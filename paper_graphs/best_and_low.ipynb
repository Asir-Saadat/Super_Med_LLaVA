```python
import matplotlib.pyplot as plt
import numpy as np

domains     = ["CT", "MRI", "Histology", "Gross Pathology", "X-ray"]
beta_acc    = [0.65, 0.66, 0.66, 0.69, 0.65]
lowest_acc  = [0.96, 0.93, 0.93, 0.89, 0.93]

x = np.arange(len(domains))

fig, ax = plt.subplots(figsize=(5, 5))

# Vertical connector for each domain
for i in range(len(domains)):
    ax.plot([x[i], x[i]], [beta_acc[i], lowest_acc[i]],
            color='gray', linewidth=1.5, zorder=1)

# Beta — solid filled dots
ax.scatter(x, beta_acc,   color='steelblue', s=80, zorder=3,
           label='Beta(2,5) Sampling')

# Lowest — hollow dots
ax.scatter(x, lowest_acc, facecolors='none', edgecolors='tomato',
           s=80, linewidths=2, zorder=3, label='Lowest Similarity')

# Random chance line
ax.axhline(0.5, color='black', linestyle='--', linewidth=1.2, label='Random Chance (0.5)')

ax.set_xticks(x)
ax.set_xticklabels(domains, fontsize=9)
ax.set_ylabel("Text-Only Classification Accuracy", fontsize=11)
ax.set_xlabel("Domain", fontsize=11)
ax.set_ylim(0.0, 1.05)
ax.legend(fontsize=10)
ax.set_title("Textual Bias: Beta Sampling vs Lowest Similarity Negatives", fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig("/home/as5606/projects/Super_LlaVA_Med/paper_graphs/bias_comparison.pdf", bbox_inches="tight")
plt.show()
```
